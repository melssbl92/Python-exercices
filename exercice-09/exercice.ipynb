```python
import pandas as pd
titanic = pd.read_csv("../data/titanic_propre.csv")
titanic["survived"].astype(int)
```

```python
survie_sexe_classe = pd.pivot_table(titanic, values="survived" , index="pclass", columns="sex", aggfunc="mean").round(3)
survie_sexe_classe
```

oui dans les 3 classes elles ont mieux survécu que les hommes !!

```python
titanic["tranche_age"] = pd.cut(titanic["age"], bins=[0, 12, 18, 60, 100],labels=["enfant", "ado", "adulte", "senior"])
survie_tranche_age = titanic.groupby("tranche_age", observed=True)["survived"].mean().round(3)
print(survie_tranche_age)
```

yes les enfants ont la priorité, surtout de 0-12 ans ils ont un meilleur taux de survie, les ados s'en sortent moins bien 

```python
prix_moyen_port = titanic.groupby("embark_town", observed=True)["fare"].mean().round(2)
print(prix_moyen_port)
repartition_classe_port = pd.crosstab(titanic["embark_town"], titanic["pclass"], normalize="index").round(3)
print(repartition_classe_port)
```

cest Cherbourg qui a le prix moyen le plus élevé
c'est le port ou les passagers utilisent le + la 1ère classe  

```python
survie_seul = (titanic["sibsp"] + titanic["parch"] == 0)
print(survie_seul)
```

voyager seul diminuait les chances de survie
