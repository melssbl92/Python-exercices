```python
import pandas as pd 
titanic = pd.read_csv("../data/titanic_propre.csv")
titanic["survived"] = titanic["survived"].astype(int)
ports = pd.DataFrame({ "embark_town": ["Southampton" , "Cherbourg", "Queenstown"], "pays" : ["Anglettere", "France", "Irlande"],})
```

```python
nb_avant = len(titanic)
titanic = titanic.merge(ports, on="embark_town", how="left", validate="many_to_one")
nb_apres = len(titanic)
assert nb_apres == nb_avant
print(nb_avant, nb_apres)
```

```python
effectifs_par_pays = titanic["pays"].value_counts()
print(effectifs_par_pays)
taux_survie_par_pays = titanic.groupby("pays")["survived"].mean().round(3)
print(taux_survie_par_pays)
```

```python
ports_sans_queenstown = ports[ports["embark_town"] != "Queenstown"]
base = titanic.drop(columns="pays")
jointure_inner= base.merge(ports_sans_queenstown, on="embark_town", how="inner")
jointure_left= base.merge(ports_sans_queenstown, on="embark_town", how="left")
print("inner:", len(jointure_inner))
print("left:", len(jointure_left))
print("pays manquants (left):", jointure_left["pays"].isna().sum())
```

pour inner il y a que les lignes dont le port existe dans 'ports' qui sont gardées, il n'y a plus les passagers de queenstown donc 77 lignes
pour left toutes les lignes de titanic sont gardées il y a tjrs les passagers de queenstown mais leur pays vaut nan
